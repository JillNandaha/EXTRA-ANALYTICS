```python
#data cleaning
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(r"[^\w\s]", "", regex=True)
    .str.replace(r"\s+", "_", regex=True)
)
```

```python
df['verdict'] = df['verdict'].replace(
    {
        'guilty': 'guilty',
    'not guilty': 'not_guilty',
    }
)
```

```python
df['court_date_first_appearance'] = pd.to_datetime(df['court_date_first_appearance'], errors='coerce', format='mixed')
df['date_of_murder'] = pd.to_datetime(df['date_of_murder'], errors='coerce', format='mixed')
df['published_date'] = pd.to_datetime(df['published_date'], errors='coerce', format='mixed')
df['verdict_date'] = pd.to_datetime(df['verdict_date'], errors='coerce', format='mixed')
```

```python
df['age'].fillna(df['age'].median(), inplace=True)
df['county'].fillna('Unknown', inplace=True)
```

```python
df['year']= df['published_date'].dt.year
```
